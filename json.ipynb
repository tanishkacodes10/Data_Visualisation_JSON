```python
#create a dataframe
import pandas as pd
sales={"product":["Pen","Notebook","Pen","Notebook"],
                    "units":[10,20,10,40],
                    "unit_price":[10,50,10,50]}
sales=pd.DataFrame(sales)
print(sales)
```

```python
#creating a new column
sales["Revenue"]=sales["units"]*sales["unit_price"]
print(sales)
```

```python
sales.to_csv("sales_report.csv", index=False)
print(sales)
```

```python
print(sales.groupby("product")["Revenue"].sum())
```

```python
#writing in tsv format
sales.to_csv("clean_data1.tsv",sep="\t",index=False)
print(sales)
```
